```python
# Logistic Regression From Scratch
```

```python
import numpy as np

def sigmoid(z):
    return 1 / (1 + np.exp(-z))
```

```python
print(sigmoid(0))
print(sigmoid(2))
print(sigmoid(-2))
print(sigmoid(np.array([-3, -2, -1, 0, 1, 2, 3])))
```

```python
def log_loss(y_true, y_pred):
    epsilon = 1e-15
    
    y_pred = np.clip(y_pred, epsilon, 1 - epsilon)
    
    loss = -np.mean(
        y_true * np.log(y_pred) +
        (1 - y_true) * np.log(1 - y_pred)
    )
    
    return loss
```

```python
y_true = np.array([1, 1, 0, 0])

good_predictions = np.array([0.9, 0.8, 0.1, 0.2])

bad_predictions = np.array([0.1, 0.2, 0.9, 0.8])

print("Good loss:", log_loss(y_true, good_predictions))
print("Bad loss :", log_loss(y_true, bad_predictions))
```

```python
class LogisticRegressionScratch:

    def __init__(self, learning_rate=0.01, epochs=1000):
        self.learning_rate = learning_rate
        self.epochs = epochs
        self.weights = None
        self.bias = None
        self.loss_history = []

    def fit(self, X, y):

        n_samples, n_features = X.shape

        self.weights = np.zeros(n_features)
        self.bias = 0

        for epoch in range(self.epochs):

            # Linear prediction
            z = np.dot(X, self.weights) + self.bias

            # Convert to probability
            y_pred = sigmoid(z)

            # Calculate loss
            loss = log_loss(y, y_pred)
            self.loss_history.append(loss)

            # Calculate gradients
            dw = (1 / n_samples) * np.dot(
                X.T,
                (y_pred - y)
            )

            db = (1 / n_samples) * np.sum(
                y_pred - y
            )

            # Update weights
            self.weights -= self.learning_rate * dw
            self.bias -= self.learning_rate * db

        return self

    def predict_probability(self, X):

        z = np.dot(X, self.weights) + self.bias

        return sigmoid(z)

    def predict(self, X, threshold=0.5):

        probabilities = self.predict_probability(X)

        return (probabilities >= threshold).astype(int)
```

```python
from collections import Counter

class_counts = Counter(y_train_np)

n_samples = len(y_train_np)
n_classes = len(class_counts)

class_weights = {}

for class_label, count in class_counts.items():
    class_weights[class_label] = n_samples / (n_classes * count)

print("Class counts:")
print(class_counts)

print("\nClass weights:")
print(class_weights)
```

```python
class WeightedLogisticRegressionScratch:
    def __init__(self, learning_rate=0.01, epochs=1000, class_weights=None):
        self.learning_rate = learning_rate
        self.epochs = epochs
        self.class_weights = class_weights
        self.weights = None
        self.bias = None
        self.loss_history = []

    def fit(self, X, y):
        n_samples, n_features = X.shape

        self.weights = np.zeros(n_features)
        self.bias = 0

        sample_weights = np.array([
            self.class_weights[int(label)]
            for label in y
        ])

        for epoch in range(self.epochs):

            z = np.dot(X, self.weights) + self.bias
            y_pred = sigmoid(z)

            # Weighted error
            error = (y_pred - y) * sample_weights

            # Weighted gradients
            dw = (1 / n_samples) * np.dot(X.T, error)
            db = (1 / n_samples) * np.sum(error)

            self.weights -= self.learning_rate * dw
            self.bias -= self.learning_rate * db

            # Weighted log-loss
            epsilon = 1e-15
            y_pred_clipped = np.clip(
                y_pred,
                epsilon,
                1 - epsilon
            )

            loss = -np.mean(
                sample_weights * (
                    y * np.log(y_pred_clipped) +
                    (1 - y) * np.log(1 - y_pred_clipped)
                )
            )

            self.loss_history.append(loss)

        return self

    def predict_probability(self, X):
        z = np.dot(X, self.weights) + self.bias
        return sigmoid(z)

    def predict(self, X, threshold=0.5):
        probabilities = self.predict_probability(X)
        return (probabilities >= threshold).astype(int)
```

```python
weighted_model = WeightedLogisticRegressionScratch(
    learning_rate=0.01,
    epochs=1000,
    class_weights=class_weights
)

weighted_model.fit(X_train_processed, y_train_np)
```

```python
weighted_probability = weighted_model.predict_probability(
    X_test_processed
)

weighted_pred = weighted_model.predict(
    X_test_processed
)
```

```python
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score
)

weighted_accuracy = accuracy_score(y_test_np, weighted_pred)
weighted_precision = precision_score(y_test_np, weighted_pred)
weighted_recall = recall_score(y_test_np, weighted_pred)
weighted_f1 = f1_score(y_test_np, weighted_pred)
weighted_roc_auc = roc_auc_score(
    y_test_np,
    weighted_probability
)

print("Weighted Model Evaluation Results")
print("----------------------------------")
print("Accuracy :", weighted_accuracy)
print("Precision:", weighted_precision)
print("Recall   :", weighted_recall)
print("F1-Score :", weighted_f1)
print("ROC-AUC  :", weighted_roc_auc)
```
